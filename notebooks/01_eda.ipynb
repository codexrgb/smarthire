# SmartHire - Exploratory Data Analysis (EDA)
This notebook inspects the raw and processed resume and job listing datasets.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import sys

sys.path.append('..')
from src.config import PROCESSED_DATA_DIR

# Load processed datasets
resumes_df = pd.read_csv(PROCESSED_DATA_DIR / 'processed_resumes.csv')
jobs_df = pd.read_csv(PROCESSED_DATA_DIR / 'processed_jobs.csv')

print(f'Resumes dataset shape: {resumes_df.shape}')
print(f'Jobs dataset shape: {jobs_df.shape}')
```

```python
# Resume Category Counts
plt.figure(figsize=(10, 5))
resumes_df['category'].value_counts().plot(kind='bar', color='skyblue')
plt.title('Resume Distribution across Job Categories')
plt.ylabel('Number of Resumes')
plt.xticks(rotation=45, ha='right')
plt.show()
```
